```python
#import key and initialize the database from the firebase

import firebase_admin
from firebase_admin import credentials
from firebase_admin import db
cred = credentials.Certificate(r"C:\Users\HP\Downloads\.json")


firebase_admin.initialize_app(cred,{"databaseURL":"https://-"
                                    "default-rtdb.firebaseio.com"})
```

```python
class Squirrel_data(object):
    """Returns a dictionary in JSON format"""

    def __init__(self, species, color, x, y):
        self.species = species
        self.color = color
        self.x = x
        self.y = y
        self.key = None

    def to_dict(self):
        """Convert object to dictionary"""
        return {
            "species": self.species,
            "color": self.color,
           "Coords": {"x": self.x, "y": self.y},
            "key": self.key
        }


    def upload(self):
        ref = db.reference("/Squirrels")
        keyref = ref.push(self.to_dict())
        self.key = keyref.key


    def update(self, new_data):
        if self.key:
            ref = db.reference("/Squirrels/" + self.key)
            ref.update(new_data)
        else:
            print("Data has not been uploaded to Firebase yet. Call upload() first.")




#Example:
squirrel = Squirrel_data("Gray Squirrel", "Gray", 45.0, -75.0)
squirrel.upload()  # This sets squirrel.key


 

```
